# Download football-data data off their site

- curl -L -s follows redirects (-L) and is silent (-s), meaning it won't show progress or error messages. The output is redirected to the specified file using -o.
-

```python
%%bash 

download_football_data() {

    # Project directory
    PROJECT_DIR="/accounts/masters/gautierep/footy_prediction/footy-prediction"

    # Where raw Football-Data files will be stored
    RAW_DIR="$PROJECT_DIR/data/raw/football_data"

    # Football-Data base URL
    BASE_URL="https://www.football-data.co.uk/mmz4281"

    # Create directory
    mkdir -p "$RAW_DIR"

    # Download seasons from 2010/11 to 2025/26
    for year in {10..25}
    do
        next=$((10#$year + 1))

        # Convert numbers to two digits
        year=$(printf "%02d" "$year")
        next=$(printf "%02d" "$next")

        # Season name, e.g. 1011
        season="${year}${next}"

        URL="$BASE_URL/${season}/E0.csv"
        OUTPUT="$RAW_DIR/E0_${season}.csv"

        echo "Downloading Premier League ${year}/${next}..."

        curl -fL --retry 3 "$URL" -o "$OUTPUT"

        echo "Saved: $OUTPUT"
    done

    echo "Download complete."
}

download_football_data
```
